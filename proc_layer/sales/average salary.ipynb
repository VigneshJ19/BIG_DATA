```python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 22],
    "Salary": [50000, 60000, 70000, 80000, 45000],
    "Department": ["HR", "Engineering", "Marketing", "Engineering", "HR"]
})

# Display the DataFrame
print("Original DataFrame:")
print(df)

# Basic statistics
print("\nBasic Statistics:")
print(df.describe(include='all'))

# Filter employees with salary greater than 55000
high_salary = df[df["Salary"] > 55000]
print("\nEmployees with Salary > 55000:")
print(high_salary)

# Group by Department and calculate average salary
avg_salary_by_dept = df.groupby("Department")["Salary"].mean().reset_index()
avg_salary_by_dept.columns = ["Department", "Average Salary"]
print("\nAverage Salary by Department:")
print(avg_salary_by_dept)
```
